20. Введите общее количество выбросов


```python
q1, q3 = np.quantile(
    variation_series, [0.25, 0.75], method="linear"
)

iqr = q3 - q1
lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

# Для каждого элемента получаем True, если это выброс
is_outlier = (variation_series < lower) | (variation_series > upper)

print(np.count_nonzero(is_outlier))  # 4
print(variation_series[is_outlier])
# [75.37375 94.1925 104.82125 285.2375]
```

```python
# Знак ~ меняет True на False и наоборот:
# оставляем элементы, которые не являются выбросами
clean = variation_series[~is_outlier]

print(clean.size)  # 266

# Гистограмма
mlp.figure(figsize=(7, 4))
mlp.hist(clean, bins="auto", edgecolor="black")
mlp.title("Гистограмма без NA и выбросов")
mlp.xlabel("Приращение стоимости акций")
mlp.ylabel("Частота")
mlp.tight_layout()
mlp.savefig("histogram.png", dpi=200)
mlp.show()

# Ящик с усами
mlp.figure(figsize=(5, 5))
mlp.boxplot(clean, whis=(0, 100))
mlp.title("Ящик с усами без NA и выбросов")
mlp.ylabel("Приращение стоимости акций")
mlp.tight_layout()
mlp.savefig("boxplot.png", dpi=200)
mlp.show()
```
